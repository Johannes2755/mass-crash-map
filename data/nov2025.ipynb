```python
from IPython.display import display
import matplotlib.pyplot as plt
import requests
import pandas as pd
import numpy as np


all_features = []


base_url = f"https://gis.crashdata.dot.mass.gov/arcgis/rest/services/MassDOT/MASSDOT_ODP_OPEN_2025/FeatureServer/0/query"
    
params = {
    "where": " CITY_TOWN_NAME = 'CAMBRIDGE' ",
    "outFields": "*",
    "outSR": "4326",
    "f": "json",
    "returnGeometry": "true",
    "resultOffset": 0,
    "resultRecordCount": 2000
    }

while True:
        
    response = requests.get(base_url, params=params)
    data = response.json()
    features = data.get("features", [])
        
    if not features:
        break
        
    all_features.extend(features)
    params["resultOffset"] += params["resultRecordCount"]

records = [f["attributes"] for f in all_features]

massdot_df = pd.DataFrame(records)
print("Shape:", massdot_df.shape)
pd.set_option('display.max_columns', None)
massdot_df.head()
```

```python
massdot_df["date"] = pd.to_datetime(massdot_df["CRASH_DATE_TEXT"], format="%m %d %Y").dt.date
massdot_df["time"] = pd.to_datetime(massdot_df["CRASH_TIME_2"], format="%I:%M %p").dt.strftime("%H:%M:%S")
massdot_df["injuryseverity"] = np.select(
    [pd.to_numeric(massdot_df["NUMB_FATAL_INJR"], errors="coerce").fillna(0) > 0,
    pd.to_numeric(massdot_df["NUMB_NONFATAL_INJR"], errors="coerce").fillna(0) > 0
    ],
    [2, 1],
    default=0
)

nm = massdot_df["NON_MTRST_TYPE_CL"].fillna("").str.lower()
massdot_df["cyclists_involved"] = nm.str.count("bicyclist")
massdot_df["pedestrians_involved"] = nm.str.count("pedestrian")

massdot_df["source"] = "MassDOT"
massdot_df["id"] = massdot_df["CRASH_NUMB"]

def clean_street(s):
    return (s.astype("string").str.strip()
             .str.replace(r"\s+Rte\s*\S*", "", regex=True)   
             .str.replace(r"\s+/\s+", " & ", regex=True))

num = (pd.to_numeric(massdot_df["STREET_NUMB"], errors="coerce")
         .astype("Int64").astype("string"))
rdwy = clean_street(massdot_df["RDWY"])
near = clean_street(massdot_df["NEAR_INT_RDWY"])
massdot_df["police"] = (massdot_df["POLC_AGNCY_TYPE_DESCR"].str.strip()
                  .replace({"Local police": "Local", "State police": "State"}))
addr = rdwy.copy()
addr = addr.mask(near.notna() & num.isna(), rdwy + " & " + near)
addr = addr.mask(num.notna(), num + " " + rdwy)
massdot_df["address"] = addr.fillna("UNKNOWN") + " Somerville, MA"
massdot_df["address"] = massdot_df["address"].str.replace(r"^(\d+)\s+\1\s+", r"\1 ", regex=True)
road_text = (massdot_df[["RDWY", "NEAR_INT_RDWY", "MM_RTE", "EXIT_RTE"]]
             .fillna("").agg(" ".join, axis=1))
massdot_df["interstate"] = road_text.str.contains(r"\bI[- ]?\d{2,3}\b", case=False, regex=True)
massdot_df["interstate"] = massdot_df["interstate"].map({True: int(1), False: None})
massdot_df["latitude"] = massdot_df["LAT"].round(4)
massdot_df["longitude"] = massdot_df["LON"].round(4)

massdot_full_clean = massdot_df[["source","id","address","date","injuryseverity",
                                               "time","police","cyclists_involved",
                                               "pedestrians_involved","latitude","longitude","interstate"]]

massdot_full_clean = massdot_full_clean.dropna(subset=["latitude", "longitude"]).reset_index(drop=True)

massdot_nov2025 = massdot_full_clean[
    (pd.to_datetime(massdot_full_clean["date"]).dt.year == 2025) &
    (pd.to_datetime(massdot_full_clean["date"]).dt.month == 11)
]

somerville_nov2025 = pd.read_csv("cambridge_clean_nov25_v2.csv")
massdot_nov2025.to_csv("massdot_nov2025.csv", index = False)

combined_nov2025 = pd.concat(
    [somerville_nov2025, massdot_nov2025],
    ignore_index=True
)

display(combined_nov2025.head())
display(combined_nov2025.tail())
```

# De-duplication Functions

```python
def within_100_meters(lat1, lon1, lat2, lon2):
    R = 6371000
    distance = 2 * R * np.arcsin(np.sqrt(np.sin((np.radians(lat2) - np.radians(lat1)) / 2) ** 2 + 
                np.cos(np.radians(lat1)) * np.cos(np.radians(lat2)) * np.sin((np.radians(lon2) - 
                np.radians(lon1)) / 2) ** 2))

    return distance <= 100

def within_6_hours(dt1, dt2):
    time_diff = abs(dt1 - dt2)
    return time_diff <= pd.Timedelta("6 hours")


def find_duplicates(df):
    df["datetime"] = pd.to_datetime(
        df["date"].astype("string") + " " + df["time"].astype("string"),
        errors="coerce"
    )

    df = df.sort_values("datetime").reset_index(drop=True)

    times = df["datetime"]

    crashes = df.to_dict(orient="records")

    duplicates = []

    for i in range(len(crashes)):
        j = i + 1
        while j < len(crashes) and within_6_hours(times[i], times[j]):
            if (
                crashes[i]["source"] != crashes[j]["source"]
                and within_100_meters(
                    crashes[i]["latitude"], crashes[i]["longitude"],
                    crashes[j]["latitude"], crashes[j]["longitude"]
                )
            ):
                duplicates.append((
                    crashes[i]["id"], crashes[j]["id"]
                ))
            j += 1

    if len(duplicates) == 0:
        return pd.DataFrame()

    pairs = pd.DataFrame(duplicates, columns=["id_1", "id_2"])

    cols = ["source", "id", "datetime", "address", "latitude", "longitude"]
    info = df[cols].drop_duplicates("id")

    out = (
        pairs
        .merge(info.add_suffix("_1"), on="id_1", how="left")
        .merge(info.add_suffix("_2"), on="id_2", how="left")
    )

    return out
```

# De-duplication on Somerville and MassDOT Data

```python
duplicates_combined = find_duplicates(combined_nov2025)
print(f"{len(duplicates_combined)} potential duplicates found")
duplicates_combined
```
