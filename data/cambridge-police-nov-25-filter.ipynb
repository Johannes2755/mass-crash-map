# Filter Cambridge Police crashes: November 2025

Load `cambridge-police-regularized.csv`, keep only records dated from November 1 through November 30, 2025, and save the unchanged columns to `cambridge-police-nov-25.csv`.

```python
from pathlib import Path

import pandas as pd

DATA_DIR = Path.cwd()
if not (DATA_DIR / "cambridge-police-regularized.csv").is_file():
    DATA_DIR = DATA_DIR / "data"

INPUT_FILE = DATA_DIR / "cambridge-police-regularized.csv"
OUTPUT_FILE = DATA_DIR / "cambridge-police-nov-25.csv"

if not INPUT_FILE.is_file():
    raise FileNotFoundError(f"Input CSV not found: {INPUT_FILE}")

crashes = pd.read_csv(INPUT_FILE)
if "date" not in crashes.columns:
    raise ValueError("Input CSV must contain a 'date' column")

crash_dates = pd.to_datetime(crashes["date"], errors="coerce")
november_2025 = crashes.loc[
    crash_dates.ge("2025-11-01") & crash_dates.lt("2025-12-01")
].copy()

november_2025.to_csv(OUTPUT_FILE, index=False)
print(f"November 2025 crash rows: {len(november_2025):,}")
print(f"Saved filtered data to {OUTPUT_FILE}")
display(november_2025.head())
```
